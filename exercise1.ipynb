## Q1. Reverse string and every 2nd character

```python
def string_ops(s):
    reversed_str = s[::-1]
    every_2nd = s[1::2]   # 2nd, 4th, 6th ... characters
    return reversed_str, every_2nd

print(string_ops("Programming"))
```

## Q2. Squares of even numbers (list comprehension)

```python
def even_squares(nums):
    return [n**2 for n in nums if n % 2 == 0]

print(even_squares([1, 2, 3, 4, 5, 6]))
```

## Q3. Character frequency using a dictionary

```python
def char_frequency(s):
    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print(char_frequency("hello world"))
```

## Q4. Two lists to dictionary using zip()

```python
def lists_to_dict(keys, values):
    return dict(zip(keys, values))

print(lists_to_dict(["a", "b", "c"], [1, 2, 3]))
```

## Q5. iter() and next() with StopIteration handling

```python
my_list = [10, 20, 30, 40]
it = iter(my_list)

while True:
    try:
        print(next(it))
    except StopIteration:
        print("End of list reached (StopIteration caught)")
        break
```

## Q6. Palindrome check using slicing

```python
def is_palindrome(s):
    return s == s[::-1]

print(is_palindrome("madam"))
print(is_palindrome("hello"))
```

## Q7. Second largest number without sorting

```python
def second_largest(nums):
    largest = second = None
    for n in nums:
        if largest is None or n > largest:
            second = largest
            largest = n
        elif n != largest and (second is None or n > second):
            second = n
    return second

print(second_largest([5, 1, 9, 7, 9, 3]))
```
